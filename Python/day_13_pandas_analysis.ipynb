# Dzień 13 - Pandas: podstawowa analiza danych
Powtórzenie filtrowania oraz nauka sortowania, podstawowych agregacji i tworzenia nowych kolumn.

```python
import pandas as pd
produkty = {
    "produkt": ["Laptop", "Monitor", "Klawiatura", "Mysz"],
    "kategoria": ["Komputery", "Komputery", "Akcesoria", "Akcesoria"],
    "cena": [3499.99, 899.99, 249.90, 129.99],
    "liczba_sztuk": [2, 5, 8, 10]
}
df = pd.DataFrame(produkty)
```

## Zadanie 1 - Powtórka podstaw Pandas

```python
df.head()
df.shape
df[["produkt", "cena"]]
df[df["cena"] > 500]
```

## Zadanie 2 - Sortowanie danych

```python
df.sort_values("cena", ascending=False)
```

## Zadanie 3 - Podstawowe obliczenia

```python
df["cena"].mean()
df["liczba_sztuk"].sum()
df["cena"].max()
df["cena"].min()
```

## Zadanie 4 - Obliczanie wartości sprzedaży

```python
df["wartosc_sprzedazy"]=df["cena"] * df["liczba_sztuk"]
df
```

## Zadanie 5 - Filtrowanie według wartości sprzedaży

```python
df[df["wartosc_sprzedazy"]> 2000]
```

## Zadanie 6 - Filtrowanie i sortowanie danych

```python
df[df["wartosc_sprzedazy"] > 1500].sort_values("wartosc_sprzedazy", ascending=False)
```

## Zadanie 7 - Analiza wartości sprzedaży

```python
df[df["kategoria"] == "Akcesoria"].sort_values("wartosc_sprzedazy", ascending=False)
```

## Zadanie 8 - Produkt z najwyższą wartością sprzedaży

```python
df["wartosc_sprzedazy"].max()
df.sort_values("wartosc_sprzedazy", ascending=False)
```
