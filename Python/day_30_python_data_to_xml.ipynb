## Zadanie 1 - Tworzenie XML na podstawie słownika

```python
dane_produktu = {
    "id": 1,
    "nazwa": "Laptop",
    "cena": 3499.99
}

import xml.etree.ElementTree as ET
root = ET.Element("produkty")
produkt = ET.SubElement(root, "produkt")
nazwa = ET.SubElement(produkt, "nazwa")
nazwa.text = dane_produktu["nazwa"]
cena = ET.SubElement(produkt, "cena")
cena.text = str(dane_produktu["cena"])
```
