## Zadanie 1 - Powtórka funkcji
Obliczenie zysku na podstawie ceny sprzedaży, kosztu jednostkowego i liczby sprzedanych sztuk.

```python
def oblicz_zysk(cena, koszt, liczba_sztuk):
    wynik = (cena * liczba_sztuk) - (koszt * liczba_sztuk)
    return wynik
zysk = oblicz_zysk(189.99, 125, 14)
print(f"Zysk: {zysk:.2f} zł ")
```

## Zadanie 2 - Realizacja celu sprzedażowego
Sprawdzenie, czy miesięczna sprzedaż osiągnęła założony cel.

```python
sprzedaz = 12500
cel = 10000

if sprzedaz >= cel:
    print("Cel sprzedażowy osiągnięty!")
else:
    print("Cel sprzedażowy nie został osiągnięty.")
```

## Zadanie 3 - Klasyfikacja wyniku sprzedaży
Klasyfikacja sprzedaży w zależności od stopnia realizacji celu.

```python
sprzedaz = 9200
cel = 10000
if sprzedaz >= cel:
    print("Cel osiągnięty!")
elif sprzedaz >= 0.8 * cel:
    print("Blisko celu!")
else:
    print("Cel nieosiągnięty!")
```

## Zadanie 4 - Warunki z operatorem logicznym
Sprawdzenie, czy pracownik spełnia wszystkie warunki przyznania premii.

```python
sprzedaz = 12500
ocena_klienta = 4.7
if sprzedaz >= 10000 and ocena_klienta >= 4.5:
    print("Premia przyznana!")
else:
    print("Premia nie została przyznana.")
```

## Zadanie 5 - Operator 'or'
Identyfikacja zamówień wymagających dodatkowej uwagi.

```python
wartosc_zamowienia = 3200
reklamacja = True

if wartosc_zamowienia > 5000 or reklamacja:
    print("Zamówienie wymaga uwagi.")
else:
    print("Zamówienie standardowe.")
```

## Zadanie 6 - Operator 'not'
Sprawdzenie, czy klient ma możliwość złożenia zamówienia.

```python
klient_zablokowany = False
if not klient_zablokowany:
    print("Klient może złożyć zamówienie.")
else:
    print("Klient nie może złożyć zamówienia.")
```

## Zadanie 7 - Klasyfikacja zamówienia
Klasyfikacja zamówienia na podstawie kilku warunków biznesowych.

```python
wartosc_zamowienia = 7500
ocena_klienta = 4.8
reklamacja = False
klient_zablokowany = False

if klient_zablokowany == True or reklamacja == True:
    print("Zamówienie wymaga weryfikacji.")
elif wartosc_zamowienia >= 5000 and ocena_klienta >= 4.5:
    print("Klient VIP.")
else:
    print("Standardowe zamówienie.")
```
