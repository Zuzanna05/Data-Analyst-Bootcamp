## Zadanie 1 - Iterowanie po liście
Wyświetlanie kolejnych elementów listy za pomocą pętli 'for'.

```python
produkty = ["Laptop", "Monitor", "Klawiatura", "Mysz", "Słuchawki"]
for produkt in produkty:
    print(produkt) 
```

## Zadanie 2 - Filtrowanie danych w pętli
Połączenie pętli 'for' z instrukcją warunkową 'if'.

```python
sprzedaz = [3200, 7500, 4100, 9200, 1800]
for wartosc in sprzedaz:
    if wartosc > 5000:
        print(wartosc)
```

## Zadanie 3 - Klasyfikacja wartości w pętli
Klasyfikowanie każdej wartości sprzedaży za pomocą 'for', 'if' i 'else'.

```python
sprzedaz = [3200, 7500, 4100, 9200, 1800]
for wartosc in sprzedaz:
    if wartosc >= 5000:
        print(wartosc, "- Wysoka sprzedaż")
    else:
        print(wartosc, "- Niska sprzedaż")
```

## Zadanie 4 - Pętla z 'range()'
Wykorzystanie 'range()' do wykonywania operacji określoną liczbę razy.

```python
for i in range(1,6):
    print(i)
```

## Zadanie 5 - Obliczenia w pętli z 'range()'
Wykorzystanie 'range()' do wykonywania obliczeń dla kolejnych wartości.

```python
for miesiac in range (1, 6):
    zamowienia = miesiac * 100
    print("Miesiąc ", miesiac, "- zamoówienia: ", zamowienia)
```

## Zadanie 6 - Warunek w pętli z 'range()'
Połączenie 'range()', obliczeń i instrukcji warunkowej.

```python
for miesiac in range (1,6):
    zamowienia = miesiac * 100
    if zamowienia >= 300:
        print(f"Miesiąc {miesiac} - zamówienia: {zamowienia}") 
```

## Zadanie 7 - List comprehension
Tworzenie nowej listy na podstawie istniejącej listy.

```python
sprzedaz = [3200, 7500, 4100, 9200, 1800]
podwojona_sprzedaz = [wartosc * 2 for wartosc in sprzedaz]
print(podwojona_sprzedaz)
```

## Zadanie 8 - list comprehension + warunek
Filtrowanie elementów listy za pomocą warunku 'if' w list comprehension.

```python
sprzedaz = [3200, 7500, 4100, 9200, 1800]
wysoka_sprzedaz = [wartosc for wartosc in sprzedaz if wartosc >= 5000]
print(wysoka_sprzedaz)
```

## Zadanie 9 - Analiza wyników sprzedaży
Wykorzystanie poznanych elementów Pythona do prostej analizy danych.

```python
sprzedaz = [3200, 7500, 4100, 9200, 1800, 6300, 2500]
nowa_sprzedaz = [wartosc * 1.1 for wartosc in sprzedaz if wartosc >= 5000]
print(nowa_sprzedaz)
```

## Zadanie 10 - Mini analiza sprzedaży
Samodzielne wykorzystanie list, pętli, warunków i list comprehension do analizy danych sprzedażowych.

```python
sprzedaz = [4200, 8100, 12000, 6700, 2900, 9500]
for wartosc in sprzedaz:
    if wartosc >= 5000:
        print(f"{wartosc} - Dobry wynik ")
    else:
        print(f"{wartosc} - Słaby wynik")
dobre_wyniki = [wartosc for wartosc in sprzedaz if wartosc >= 5000]
print("Nowa lista: ", dobre_wyniki)

srednia=sum(dobre_wyniki)/len(dobre_wyniki)
print("Średnia: ", srednia)
```
