## Zakres

- czym są brakujące wartości 'NaN'
- wykrywanie braków za pomocą 'isna()'
- liczenie brakujących wartości za pomocą 'isna().sum()'
- uzupełnianie braków za pomocą 'fillna()'
- podejmowanie decyzji, kiedy brak danych można zastąpić konkretną wartością

## Zadanie 1 - Powtórka merge() i LEFT JOIN

```python
import pandas as pd
produkty = {
    "id": [1, 2, 3, 4],
    "produkt": ["Laptop", "Monitor", "Klawiatura", "Mysz"],
    "cena": [3499.99, 899.99, 249.90, 129.99]
}

zamowienia = {
    "produkt_id": [1, 2, 1, 3],
    "liczba_sztuk": [2, 5, 3, 4]
}

df_produkty = pd.DataFrame(produkty)
df_zamowienia = pd.DataFrame(zamowienia)

df_polaczone = pd.merge (
    df_produkty,
    df_zamowienia,
    left_on="id",
    right_on="produkt_id",
    how="left"
)
df_polaczone
```

## Zadanie 2 - Wykrywanie brakujących wartości

```python
df_polaczone.isna()
```

## Zadanie 3 - Liczenie brakujących wartości

```python
df_polaczone.isna().sum()
```

## Zadanie 4 - Uzupełnianie brakujących wartości

```python
df_polaczone["liczba_sztuk"] = df_polaczone["liczba_sztuk"].fillna(0)
df_polaczone.isna().sum()
```

## Zadanie 5 - Interpretacja brakujących danych

### Wniosek
Jeżeli brak wartości w "liczba_sztuk" wynika z tego, że produkt nie występuje w tabeli zamówień, można zastąpić 'NaN' wartością '0' za pomocą 'fillna(0)'.
Nie każdy brak danych należy jednak automatycznie zastępować zerem. Decyzja zależy od znaczenia danej kolumny i kontekstu danych.
