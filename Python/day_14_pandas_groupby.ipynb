# Dzień 14 - Pandas: podstawy groupby()
## Zakres
- powtórka filtrowania i sortowania
- grupowania danych za pomocą 'groupby()'
- 'groupby()' z metodą '.sum()'
- analiza sprzedaży według kategorii
- wyciąganie prostych wniosków z zagregowanych danych

```python
import pandas as pd

produkty = {
    "produkt": ["Laptop", "Monitor", "Klawiatura", "Mysz"],
    "kategoria": ["Komputery", "Komputery", "Akcesoria", "Akcesoria"],
    "cena": [3499.99, 899.99, 249.90, 129.99],
    "liczba_sztuk": [2, 5, 8, 10]
}
df = pd.DataFrame(produkty)
df["wartosc_sprzedazy"] = df["cena"] * df["liczba_sztuk"]
```

## Zadanie 1 - Powtórka z Pandas

```python
df[df["kategoria"] == "Komputery"]
df.sort_values("wartosc_sprzedazy", ascending=False)
df["cena"].mean()
```

## Zadanie 2 - Grupowanie danych według kategorii

```python
df.groupby("kategoria")["liczba_sztuk"].sum()
```

## Zadanie 3 - Łączna wartość sprzedaży według kategorii

```python
df.groupby("kategoria")["wartosc_sprzedazy"].sum()
```

Kategoria Komputery generuje znacznie większą wartość sprzedaży niż Akcesoria, mimo mniejszej liczby sprzedanych sztuk. Wynika to z dużo wyższych cen produktów w tej kategorii.

## Zadanie 4 - Średnia cena według kategorii

```python
df.groupby("kategoria")["cena"].mean()
```
