# Dzień 15 - Pandas: groupby() i analiza wyników

## Zakres
- powtórka 'groupby()'
- agregacje '.sum()', '.mean()', '.min()' i '.max'
- 'reset_index()'
- sortowanie wyników grupowania
- filtrowanie zagregowanych danych
- wyciąganie wniosków biznesowych

```python
import pandas as pd

produkty = {
    "produkt": ["Laptop", "Monitor", "Klawiatura", "Mysz"],
    "kategoria": ["Komputery", "Komputery", "Akcesoria", "Akcesoria"],
    "cena": [3499.99, 899.99, 249.90, 129.99],
    "liczba_sztuk": [2, 5, 8, 10]
}

df = pd.DataFrame(produkty)
df["wartosc_sprzedazy"] = df["cena"] * df["liczba_sztuk"]
```

## Zadanie 1 - Powtórka grupowania danych

```python
df.groupby("kategoria")["liczba_sztuk"].sum()
df.groupby("kategoria")["cena"].mean()
df.groupby("kategoria")["wartosc_sprzedazy"].sum()
```

## Zadanie 2 - Minimum i maksimum według kategorii

```python
df.groupby("kategoria")["cena"].max()
df.groupby("kategoria")["cena"].min()
```

## Zadanie 3 - Zamiana wyniku grupowania na DataFrame

```python
sprzedaz_kategorie = df.groupby("kategoria")["wartosc_sprzedazy"].sum()
sprzedaz_kategorie
sprzedaz_kategorie.reset_index()
```

## Zadanie 4 - Sortowanie zagregowanych danych

```python
sprzedaz_kategorie = sprzedaz_kategorie.reset_index()
sprzedaz_kategorie.sort_values("wartosc_sprzedazy", ascending=False)
```

## Zadanie 5 - Filtrowanie zagregowanych danych

```python
sprzedaz_kategorie[sprzedaz_kategorie["wartosc_sprzedazy"] > 5000]
```

## Zadanie 6 - Analiza sprzedaży według kategorii

```python
sprzedaz_kategorie = df.groupby("kategoria")["wartosc_sprzedazy"].sum()
sprzedaz_kategorie = sprzedaz_kategorie.reset_index()
sprzedaz_kategorie[sprzedaz_kategorie["wartosc_sprzedazy"] > 3000].sort_values("wartosc_sprzedazy", ascending=False)
```

Analiza wykazała, że kategoria Komputery generuje wyższą łączną wartość sprzedaży niż kategoria Akcesoria - 11499,93 zł w porównaniu z 3299,10 zł.
