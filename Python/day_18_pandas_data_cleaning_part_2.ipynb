## Zadanie 1 - Powtórka pracy z brakującymi danymi

```python
import pandas as pd

dane = {
    "produkt": ["Laptop", "Monitor", "Klawiatura", "Mysz"],
    "cena": [3499.99, 899.99, 249.90, 129.99],
    "liczba_sztuk": [2, None, 4, None]
}

df = pd.DataFrame(dane)
df
df.isna().sum()
```

## Zadanie 2 - Usuwanie brakujących danych za pomocą dropna()

```python
df.dropna()
```

## Zadanie 3 - Wykrywanie duplikatów

```python
dane = {
    "produkt": ["Laptop", "Monitor", "Klawiatura", "Mysz", "Monitor"],
    "cena": [3499.99, 899.99, 249.90, 129.99, 899.99]
}

df = pd.DataFrame(dane)
df

df.duplicated()
df.duplicated().sum()
```

## Zadanie 4 - Usuwanie duplikatów za pomocą drop_duplicates()

```python
df.drop_duplicates()
df
df = df.drop_duplicates()
df.duplicated().sum()
```

## Zadanie 5 - Czyszczenie braków i duplikatów

```python
dane = {
    "produkt": ["Laptop", "Monitor", "Klawiatura", "Mysz", "Mysz"],
    "cena": [3499.99, 899.99, None, 129.99, 129.99],
    "liczba_sztuk": [2, 5, 4, None, None]
}

df = pd.DataFrame(dane)
df

df.isna().sum()
df.duplicated().sum()

df= df.drop_duplicates()
df = df.dropna()
df
df.isna().sum()
df.duplicated().sum()
```
