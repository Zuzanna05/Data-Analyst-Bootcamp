## Zadanie 1 - Powtórka czyszczenia danych

```python
import pandas as pd

dane = {
    "id": [1, 2, 3, 4, 4],
    "produkt": ["Laptop", "Monitor", "Klawiatura", "Mysz", "Mysz"],
    "cena": [3499.99, 899.99, None, 129.99, 129.99],
    "liczba_sztuk": [2, 5, 4, 10, 10]
}

df = pd.DataFrame(dane)
df
df.isna().sum()
df.duplicated().sum()
```

## Zadanie 2 - Wykrywanie duplikatów według wybranej kolumny

```python
df.duplicated(subset="id")
df = df.drop_duplicates(subset="id")
```

## Zadanie 3 - Podstawowy pipeline czyszczenia danych

```python
df = df.dropna()
df.isna().sum()
df.duplicated(subset="id").sum()
```
