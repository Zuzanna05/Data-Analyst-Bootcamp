## Zadanie 1 - Podstawowe operacje na liście
Praca z listą produktów: indeksowanie i sprawdzanie długości listy.

```python
produkty = ["Laptop", "Monitor", "Klawiatura", "Mysz", "Słuchawki"]
print ("Pierwszy produkt: ", produkty[0])
print("Ostatni produkt: ", produkty[-1])
print ("Liczba produktów: ", len(produkty))
print("Trzeci produkt: ", produkty[2])
```

## Zadanie 2 - Modyfikowanie listy
Dodawanie, usuwanie i zmiana elementów listy produktów.

```python
produkty = ["Laptop", "Monitor", "Klawiatura", "Mysz"]
produkty.append("Słuchawki")
produkty.remove("Klawiatura")
produkty[1] = "Monitor 4K"
print(produkty)
```

## Zadanie 3 - Sprawdzanie elementów listy
Sprawdzenie, czy określony produkt znajduje się na liście.

```python
produkty = ["Laptop", "Monitor", "Klawiatura", "Mysz", "Słuchawki"]
print("Monitor: ", "Monitor" in produkty)
print("Tablet: ", "Tablet" in produkty)
```

## Zadanie 4 - Podstawowa analiza danych sprzedażowych
Wykorzystanie funkcji 'min()', 'max()' i 'sum()' do analizy listy wartości sprzedaży.

```python
sprzedaz = [3200, 7500, 4100, 9200, 1800]
print("Najniższa sprzedaż: ", min(sprzedaz))
print("Najwyższa sprzedaż: ", max(sprzedaz))
print("Łączna sprzedaż: ", sum(sprzedaz))
print("Liczba wyników: ", len(sprzedaz))
print("Średnia sprzedaż: ", sum(sprzedaz) / len(sprzedaz))
```
