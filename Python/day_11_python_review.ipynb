## Zadanie 1 - Filtrowanie danych
Wybranie z listy wartości sprzedaży większych lub równych 5000.

```python
sprzedaz = [3200, 7500, 4100, 9200, 1800, 6300, 2500]
wysoka_sprzedaz = [wartosc for wartosc in sprzedaz if wartosc >= 5000]
print(wysoka_sprzedaz)
```

## Zadanie 2 - Obliczanie średniej sprzedaży

```python
wysoka_sprzedaz = [7500, 9200, 6300]
srednia = sum(wysoka_sprzedaz) / len(wysoka_sprzedaz)
print(f"Średnia wynosi: {srednia}")
```

## Zadanie 3 - Funkcja obliczająca wartość sprzedaży

```python
produkt = {
    "produkt": "Laptop",
    "cena": 3499.99,
    "liczba_sztuk": 2
}
def oblicz_sprzedaz (produkt):
    wartosc_sprzedazy = produkt["cena"] * produkt["liczba_sztuk"]
    return wartosc_sprzedazy
print(f"Wartość sprzedaży wynosi: {oblicz_sprzedaz(produkt)}")
```
