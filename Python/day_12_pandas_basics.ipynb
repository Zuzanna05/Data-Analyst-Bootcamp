# Dzień 12 - Pandas: podstawy analizy danych
Pierwsze kroki z biblioteką Pandas: DataFrame, poznawanie struktury danych, wybieranie kolumn i filtrowanie.

```python
import pandas as pd
```

## Zadanie 1 - Tworzenie pierwszego DataFrame

```python
produkty = {
    "produkt": ["Laptop", "Monitor", "Klawiatura", "Mysz"],
    "kategoria": ["Komputery", "Komputery", "Akcesoria", "Akcesoria"],
    "cena": [3499.99, 899.99, 249.90, 129.99],
    "liczba_sztuk": [2, 5, 8, 10]
}
df = pd.DataFrame(produkty)
df
```

## Zadanie 2 - Podstawowe informacje o DataFrame

```python
df.head()
df.head(2)
df.shape
df.columns
df.info()
```

## Zadanie 3 - Wybieranie kolumn z DataFrame

```python
df["produkt"]
df[["produkt","cena"]]
```

## Zadanie 4 - Filtrowanie danych

```python
df["cena"] > 500
df[df["cena"] > 500]
df[df["kategoria"] == "Akcesoria"]
df[df["liczba_sztuk"] >= 5]
```
