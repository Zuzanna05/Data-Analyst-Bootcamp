## Zadanie 1 - Utworzenie przykładowego pliku XML

    <produkt id="2">

## Zadanie 2 - Wczytanie pliku XML za pomocą ElementTree

```python
import xml.etree.ElementTree as ET

tree = ET.parse("data/produkty.xml")
```

## Zadanie 3 - Pobranie elementu głównego XML

```python
root = tree.getroot()
```

## Zadanie 4 - Sprawdzanie elementu głównego XML

```python
root.tag
```

## Zadanie 5 - Mini challenge: odczyt struktury XML

```python
for element in root:
    print(element.tag)
```
