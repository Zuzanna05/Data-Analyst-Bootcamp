## Zadanie 1 - Tworzenie elementu potomnego XML

```python
import xml.etree.ElementTree as ET

root = ET.Element("produkty")
produkt = ET.SubElement(root, "produkt")
nazwa = ET.SubElement(produkt, "nazwa")
nazwa.text = "Laptop"
cena = ET.SubElement(produkt, "cena")
cena.text = "3499.99"
```
