## Zadanie 1 - Dodanie atrybutu id do elementu XML

```python
dane_produktu = {
    "id": 1,
    "nazwa": "Laptop",
    "cena": 3499.99
}

import xml.etree.ElementTree as ET

root = ET.Element("produkty")
produkt = ET.SubElement(root, "produkt")

produkt.set("id", str(dane_produktu["id"]))
produkt.set("nazwa", dane_produktu["nazwa"])
```
