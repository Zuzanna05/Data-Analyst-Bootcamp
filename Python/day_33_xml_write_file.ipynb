# Dzień 33 - XML: zapis danych do pliku

## Zakres
- powtórka tworzenia struktury XML
- utworzenie drzewa 'ElementTree'
- zapis XML do pliku
- kodowanie UTF-8
- kontrola utworzonego pliku

## Zadanie 1 - Utworzenie struktury XML z listy produktów

```python
lista_produktow = [
    {"id": 1, "nazwa": "Laptop", "cena": 3499.99},
    {"id": 2, "nazwa": "Monitor", "cena": 899.99},
    {"id": 3, "nazwa": "Klawiatura", "cena": 249.90}
]

import xml.etree.ElementTree as ET

root = ET.Element("produkty")

for produkt in lista_produktow:
    element_produktu = ET.SubElement(root,"produkt")
    element_produktu.set("id", str(produkt["id"]))
    nazwa = ET.SubElement(element_produktu,"nazwa")
    nazwa.text = produkt["nazwa"]
    cena = ET.SubElement(element_produktu, "cena")
    cena.text = str(produkt["cena"])
```

## Zadanie 2 - Utworzenie drzewa ElementTree

```python
tree = ET.ElementTree(root)
```

## Zadanie 3 - Zapis drzewa XML do pliku

```python
tree.write(
    "data/produkty_wygenerowane.xml",
    encoding="utf-8",
    xml_declaration=True
           )
```
