Zakres:

- pary klucz-wartość
- odczytywanie wartości za pomocą kluczy
- dodawanie i modyfikowanie danych
- iterowanie po słownikach
- łączenie słowników z 'for' i 'if'
- listy słowników

## Zadanie 1 - Tworzenie słownika
Przechowywanie informacji o produkcie za pomocą par klucz-wartość.

```python
produkt = {
    "produkt":"Laptop",
    "cena":3499.99,
    "liczba_sztuk":2,
    "kategoria":"Elektronika"
}
print(produkt)
```

## Zadanie 2 - Odczytywanie wartości ze słownika
Pobieranie konkretnych informacji ze słownika za pomocą kluczy.

```python
print(f"Nazwa produktu: {produkt['produkt']}")
print(f"Cena: {produkt['cena']}")
print(f"Kategoria: {produkt['kategoria']}")
```

## Zadanie 3 - Modyfikowanie słownika
Zmiana istniejących wartości oraz dodawanie nowych par klucz-wartość.

```python
produkt['cena'] = 3299.99
produkt['liczba_sztuk'] = 5
produkt['promocja'] = True
print(produkt)
```

## Zadanie 4 - Sprawdzanie danych w słowniku
Wykorzystanie wartości ze słownika w instrukcji warunkowej 'if'.

```python
if produkt['cena'] > 3000:
    print("Drogi produkt")
else:
    print("Tani produkt")
```

## Zadanie 5 - Iterowanie po słowniku
Wyświetlanie kluczy i wartości słownika za pomocą pętli 'for'.

```python
for klucz, wartosc in produkt.items():
    print(f"{klucz}:{wartosc}")
```

## Zadanie 6 - Lista słowników
Przechowywanie wielu produktów jako listy słowników i odczytywanie ich danych.

```python
produkty = [
    {"produkt": "Laptop", "cena": 3499.99, "liczba_sztuk": 2},
    {"produkt": "Monitor", "cena": 899.99, "liczba_sztuk": 5},
    {"produkt": "Klawiatura", "cena": 249.90, "liczba_sztuk": 8},
    {"produkt": "Mysz", "cena": 129.99, "liczba_sztuk": 12}
]

for produkt in produkty:
    print(produkt['produkt'])
```

## Zadanie 7 - Filtrowanie danych w liście słowników
Połączenie pętli 'for', słowników i instrukcji warunkowej 'if'.

```python
for produkt in produkty:
    if produkt['cena'] > 500:
        print(f"{produkt['produkt']} - {produkt['cena']}")
```

## Zadanie 8 - Obliczanie wartości sprzedaży
Wykorzystanie danych ze słowników do wykonania prostych obliczeń biznesowych.

```python
for produkt in produkty:
    wartosc_sprzedazy = produkt['cena'] * produkt['liczba_sztuk']
    print(f"{produkt['produkt']} - wartość sprzedaży: {wartosc_sprzedazy}")
```

## Zadanie 9 - Analiza sprzedaży produktów
Samodzielne wykorzystanie listy słowników, pętli i warunków do analizy danych sprzedażowych.

```python
for produkt in produkty:
    wartosc_sprzedazy = produkt['cena'] * produkt['liczba_sztuk']
    if wartosc_sprzedazy >= 4000:
        print(f"{produkt['produkt']} - wysoka wartość sprzedaży")
    else:
        print(f"{produkt['produkt']} - niska wartość sprzedaży")
```

## Zadanie 10 - Podsumowanie sprzedaży
Samodzielna analiza listy słowników i obliczenie łącznej wartości sprzedaży.

```python
laczna_wartosc_sprzedazy = 0
for produkt in produkty:
        wartosc_sprzedazy = produkt['cena'] * produkt['liczba_sztuk']
        laczna_wartosc_sprzedazy += wartosc_sprzedazy
print(f"Łączna wartość sprzedaży: {laczna_wartosc_sprzedazy:.2f}")
```
