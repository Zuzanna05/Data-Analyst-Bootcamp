## Zadanie 1 - Prosta funkcja

```python
def powitanie():
    print("Hello World!")

powitanie()
```

# Zadanie 2 - Funkcja z parametrem
Przekazywanie danych do funkcji za pomocą parametru.

```python
def powitanie_analityka(imie):
    print(f"Witaj {imie}")
powitanie_analityka("Anna")
```

# Zadanie 3 - Zwracanie wyniku funkcji
Wykorzystanie 'return' do zwracania obliczonej wartości z funkcji.

```python
def oblicz_wartosc_sprzedaz(cena, liczba_sztuk):
    wartosc_sprzedazy = cena * liczba_sztuk
    return wartosc_sprzedazy
oblicz_wartosc_sprzedaz(100,5)
```

## Zadanie 4 - Warunek wewnątrz funkcji
Łączenie funkcji z instrukcją warunkową 'if' w celu klasyfikacji danych.

```python
def klasyfikuj_sprzedaz(wartosc):
    if wartosc >= 5000:
        return "Wysoka sprzedaż"
    else:
        return "Niska sprzedaż"
klasyfikuj_sprzedaz(7500)
```

## Zadanie 5 - Funkcja do obliczania wartości sprzedaży
Wykorzystanie funkcji do wykonywania obliczeń na danych sprzedażowych.

```python
produkt = {
    "produkt": "Laptop",
    "cena": 3499.99,
    "liczba_sztuk": 2
}
def oblicz_sprzedaz(produkt):
    sprzedaz = produkt['cena'] * produkt['liczba_sztuk']
    return sprzedaz
oblicz_sprzedaz(produkt)
```

## Zadanie 6 - Łączenie funkcji
Wykorzystanie wyniku jednej funkcji jako danych dla kolejnej funkcji.

```python
wartosc = oblicz_sprzedaz(produkt)
print(wartosc)
print(klasyfikuj_sprzedaz(wartosc))
```

## Zadanie 7 - Funkcja i lista słowników
Wykorzystanie funkcji do analizy wielu produktów zapisanych jako lista słowników.

```python
produkty = [
    {"produkt": "Laptop", "cena": 3499.99, "liczba_sztuk": 2},
    {"produkt": "Monitor", "cena": 899.99, "liczba_sztuk": 5},
    {"produkt": "Klawiatura", "cena": 249.90, "liczba_sztuk": 8},
    {"produkt": "Mysz", "cena": 129.99, "liczba_sztuk": 12}
]
for produkt in produkty:
    wartosc = oblicz_sprzedaz(produkt)
    print(f"{produkt['produkt']} - {wartosc}")
```

## Zadanie 8 - Analiza sprzedaży za pomocą funkcji
Połączenie funkcji, listy słowników, pętli i klasyfikacji danych sprzedażowych.

```python
for produkt in produkty:
    wartosc = oblicz_sprzedaz(produkt)
    klasyfikacja = klasyfikuj_sprzedaz(wartosc)
    print(f"{produkt['produkt']} - {wartosc} - {klasyfikacja}")
```
