## Zadanie 1 - Wczytanie pliku XML i pobranie elementu głównego

```python
import xml.etree.ElementTree as ET

tree = ET.parse("data/produkty.xml")
root = tree.getroot()
```

## Zadanie 2 - Wyszukiwanie elementów za pomocą find()

```python
for produkt in root:
    nazwa = produkt.find("nazwa")
    print(nazwa)
```

## Zadanie 3 - Odczytywanie wartości elementów za pomocą .text

```python
for produkt in root:
    nazwa = produkt.find("nazwa")
    cena = produkt.find("cena")
    print(nazwa.text, " - ", cena.text)
```

## Zadanie 4 - Odczytywanie atrybutów XML

```python
id_produktu = produkt.get("id")
produkt.get("id")
```

## Zadanie 5 - Mini challenge: oczyt danych z XML

```python
for produkt in root:
    id_produktu = produkt.get("id")
    nazwa = produkt.find("nazwa")
    cena = produkt.find("cena")

    print(id_produktu, "-", nazwa.text, "-", cena.text)
```
