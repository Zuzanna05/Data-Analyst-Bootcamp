# Dzień 20 - XML: podstawowa struktura danych

## Zakres
- czym jest XML

## Zadanie 1 - Porównanie struktury CSV i XML

CSV:
id,produkt,cena
1,Laptop,3499.99
2,Monitor,899.99

XML:
<produkty>
    <produkt id="1">
        <nazwa>Laptop</nazwa>
        <cena>3499.99</cena>
    </produkt>

    <produkt id="2">
        <nazwa>Monitor</nazwa>
        <cena>899.99</cena>
    </produkt>
</produkty>

### Wniosek
CSV przechowuje dane w strukturze wierszy i kolumn, natomiast XML wykorzystuje zagnieżdżone elementy.
W dokumencie XML element '<produkty>' jest elementem głównym('root'), ponieważ zawiera wszystkie pozostałe elementy dokumentu.

## Zadanie 2 - Elementy, atrybuty i wartości tekstowe

Przykładowa struktura XML:

<klienci>
    <klient id="101">
        <imie>Anna</imie>
        <miasto>Warszawa</miasto>
    </klient>
</klienci>

- '<klienci>' - element główny ('root')
- '<klient>' - element reprezentujący pojedynczego klienta
- 'id="101"' - atrybut elementu '<klient>'
- '<imie>' i '<miasto>' - elementy zagnieżdżone
- 'Anna' i 'Warszawa' - wartości tekstowe elementów
