- rozpoznanie nieprawidłowego typu kolumny
- przygotowanie danych liczbowych do analizy

## Zadanie 1 - Kontrola typów danych w DataFrame

```python
lista_produktow = [
    {"id": "1", "nazwa": "Laptop", "cena": "3499.99"},
    {"id": "2", "nazwa": "Monitor", "cena": "899.99"},
    {"id": "3", "nazwa": "Klawiatura", "cena": "249.90"}
]

import pandas as pd
df_produkty = pd.DataFrame(lista_produktow)
df_produkty.info()
```

## Zadanie 2 - Konwersja kolumny cena na typ liczbowy

```python
df_produkty["cena"] = df_produkty["cena"].astype(float)
df_produkty.info()
```

## Zadanie 3 - Prosta analiza po konwersji typu danych

```python
srednia_cena = df_produkty["cena"].mean()
print(srednia_cena)
```
