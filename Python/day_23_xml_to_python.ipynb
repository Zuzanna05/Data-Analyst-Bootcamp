## Zadanie 1 - Wczytanie XML i powtórka odczytywania danych

```python
import xml.etree.ElementTree as ET
tree = ET.parse("data/produkty.xml")
root = tree.getroot()

for produkt in root:
     id_produktu = produkt.get("id")
     nazwa = produkt.find("nazwa")
     cena = produkt.find("cena")
     print(id_produktu, "-", nazwa.text, "-", cena.text)
```

## Zadanie 2 - Tworzenie słownika na podstawie danych XML

```python
dane_produktu = {
    "id": id_produktu,
    "nazwa": nazwa.text,
    "cena": cena.text
}
```

## Zadanie 3 - Tworzenie listy słowników z danych XML

```python
lista_produktow = []

for produkt in root:
    id_produktu = produkt.get("id")
    nazwa = produkt.find("nazwa")
    cena = produkt.find("cena")

    dane_produktu = {
        "id": id_produktu,
        "nazwa": nazwa.text,
        "cena": cena.text
    }

lista_produktow.append(dane_produktu)
print(lista_produktow)
```
