## Zadanie 1 - Powtórka groupby()

```python
import pandas as pd

produkty = {
    "produkt": ["Laptop", "Monitor", "Klawiatura", "Mysz"],
    "kategoria": ["Komputery", "Komputery", "Akcesoria", "Akcesoria"],
    "cena": [3499.99, 899.99, 249.90, 129.99],
    "liczba_sztuk": [2, 5, 8, 10]
}
df = pd.DataFrame(produkty)
df["wartosc_sprzedazy"] = df["cena"] * df["liczba_sztuk"]

sprzedaz_kategorie = df.groupby("kategoria")["wartosc_sprzedazy"].sum()
sprzedaz_kategorie = sprzedaz_kategorie.reset_index()
sprzedaz_kategorie.sort_values("wartosc_sprzedazy", ascending= False)
```

## Zadanie 2 - Przygotowanie danych do połączenia

```python
produkty = {
    "id": [1, 2, 3, 4],
    "produkt": ["Laptop", "Monitor", "Klawiatura", "Mysz"],
    "cena": [3499.99, 899.99, 249.90, 129.99]
}

zamowienia = {
    "produkt_id": [1, 2, 1, 4, 3],
    "liczba_sztuk": [2, 5, 3, 10, 4]
}

df_produkty = pd.DataFrame(produkty)
df_zamowienia = pd.DataFrame(zamowienia)
print("PRODUKTY")
display(df_produkty)
print("ZAMÓWIENIA")
display(df_zamowienia)
```

## Zadanie 3 - Łączenie DataFrame za pomocą merge()

```python
pd.merge(
    df_produkty,
    df_zamowienia,
    left_on="id",
    right_on="produkt_id"
)
```

## Zadanie 4 - Obliczanie wartości sprzedaży po połączeniu danych

```python
df_polaczone = pd.merge(
    df_produkty,
    df_zamowienia,
    left_on="id",
    right_on="produkt_id"
)

df_polaczone["wartosc_sprzedazy"] = df_polaczone["cena"] * df_polaczone["liczba_sztuk"]
df_polaczone
```

## Zadanie 5 - Łączna wartość sprzedaży według produktu

```python
sprzedaz_produkty = df_polaczone.groupby("produkt")["wartosc_sprzedazy"].sum()
sprzedaz_produkty = sprzedaz_produkty.reset_index()
sprzedaz_produkty.sort_values("wartosc_sprzedazy", ascending = False)
```

## Zadanie 6 - INNER JOIN vs LEFT JOIN

```python
nowy_produkt = pd.DataFrame({
    "id": [5],
    "produkt": ["Drukarka"],
    "cena": [699.99]
})

df_produkty = pd.concat([df_produkty, nowy_produkt], ignore_index=True)
display(df_produkty)
display(df_zamowienia)

pd.merge(
    df_produkty,
    df_zamowienia,
    left_on="id",
    right_on="produkt_id",
    how = "left"
)
```

## Zadanie 7 - Analiza sprzedaży po połączeniu danych

```python
df_raport = pd.merge (
    df_produkty,
    df_zamowienia,
    left_on = "id",
    right_on = "produkt_id",
    how = "left"
)
sprzedaz_produkty = df_raport.groupby("produkt")["liczba_sztuk"].sum()
sprzedaz_produkty = sprzedaz_produkty.reset_index()
sprzedaz_produkty.sort_values("liczba_sztuk", ascending=False)
```
