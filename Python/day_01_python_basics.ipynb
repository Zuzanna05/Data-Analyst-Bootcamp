```python
# Zadanie 2 - Mini analiza sprzedaży
laptop = 3499.99
mysz = 79.90
klawiatura = 199.90
przychod_laptopy = 3 * laptop
przychod_myszy = 12 * mysz
przychod_klawiatury = 7 * klawiatura
przychod_sklepu = przychod_laptopy + przychod_myszy + przychod_klawiatury
liczba_sprzedanych = 3 + 12 + 7
print("Przychod z laptopów: ", round(przychod_laptopy, 2))
print("Przychód z myszy: ", round(przychod_myszy, 2))
print("Przychód z klawiatur: ", round(przychod_klawiatury, 2))
print("Całkowity przychód sklepu: ", round(przychod_sklepu, 2))
print("Łączna liczba sprzedanych produktów: ", liczba_sprzedanych)
```

```python
# Zadanie 3 - Udział produktów w przychodzie
procent_laptopy = (przychod_laptopy / przychod_sklepu) * 100
procent_laptopy = round(procent_laptopy, 2)
print("Udział laptopów w przychodzie: ", procent_laptopy)
procent_myszy = (przychod_myszy / przychod_sklepu) * 100
procent_myszy = round(procent_myszy, 2)
print("Udział myszy w przychodzie: ", procent_myszy)
procent_klawiatur = (przychod_klawiatury / przychod_sklepu) * 100
procent_klawiatur = round(procent_klawiatur, 2)
print("Udział klawiatur w przychodzie: ", procent_klawiatur)

suma_udzialow = procent_laptopy + procent_myszy + procent_klawiatur
suma_udzialow = round(suma_udzialow, 2)
print("Suma udziałów: ", suma_udzialow)
```

3.Laptopy odpowiadają za 81,66% całkowitego przychodu, mimo, że sprzedano ich jedynie 3 sztuki. Może to sugerować, że są kluczową kategorią z punktu widzenia przychodów sklepu. Przed podjęciem decyzji o zwiększeniu ich udziału w ofercie należałoby jednak przeanalizować również koszty oraz marżę poszczególnych produktów. 

```python
# Zadanie 4 - Funkcje
def oblicz_przychod (cena, liczba_sztuk):
    wynik = cena * liczba_sztuk
    return wynik

przychod_laptopy = oblicz_przychod(3499.99, 3)
przychod_myszy = oblicz_przychod(79.90, 12)
przychod_klawiatury = oblicz_przychod(199.90, 7)

print(f"Przychód z laptopów: {przychod_laptopy:.2f} zł")
print(f"Przychód z myszy: {przychod_myszy:.2f} zł")
print(f"Przychód z klawiatur: {przychod_klawiatury:.2f} zł")
```
